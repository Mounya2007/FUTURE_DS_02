```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("archive (2).zip")
data.head()
```

```python
data.columns
```

```python
data.columns = data.columns.str.strip().str.lower()
data.columns
```

```python
data['churn'].value_counts()
```

```python
data['churn'].value_counts(normalize=True) * 100
```

```python
import matplotlib.pyplot as plt

data['churn'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Customer Churn Distribution")
plt.ylabel("")
plt.show()
```

```python
data.groupby('gender')['churn'].value_counts().unstack().plot(kind='bar')
plt.title("Churn by Gender")
plt.show()
```

```python
data.groupby('contract')['churn'].value_counts().unstack().plot(kind='bar')
plt.title("Churn by Contract Type")
plt.show()
```

```python
data['tenure'] = pd.to_numeric(data['tenure'], errors='coerce')

data.groupby('tenure')['churn'].value_counts().unstack().plot()
plt.title("Churn over Time")
plt.show()
```

**Insights:**
- A significant number of customers are leaving the service.
- Customers with short tenure have higher churn.
- Month-to-month contracts show higher churn rates.
- Long-term customers are more stable.
- Churn varies across different customer groups.

**Recommendations:**
- Improve onboarding for new customers
- Offer discounts for long-term contracts
- Identify high-risk customers early
- Improve customer support services
